```python
# Standardize data
# Delete row at index 0
pov_df = pov_df.drop(0)
# Reset index after deletion
pov_df = pov_df.reset_index(drop=True)
```

```python
# Split NAME column into County and State
pov_df[['County', 'State']] = pov_df['NAME'].str.split(',', expand=True)

# Clean up extra spaces
pov_df['County'] = pov_df['County'].str.strip()
pov_df['State'] = pov_df['State'].str.strip()

# Remove the word "County" from County column
pov_df['County'] = pov_df['County'].str.replace('County', '', regex=False).str.strip()

# Reorder columns: GEO_ID, year, County, State, then the rest
cols = pov_df.columns.tolist()

# Ensure order
new_order = ['GEO_ID','year', 'County', 'State'] + [col for col in cols if col not in ['GEO_ID', 'year', 'County', 'State']]

pov_df = pov_df[new_order]
pov_df = pov_df.drop('NAME', axis=1)
```

```python
# List of columns to keep
keep = [
    "GEO_ID", "NAME", "State", "County", "year",   
    "S1701_C01_001E",      # Total Population
    "S1701_C02_001E",       # Population Below Poverty
    "S1701_C03_001E",       #Percent Below Poverty
    "S1701_C01_023E",       #Less Than High School Graduate
    "S1701_C01_026E",       #Bachelor's degree or higher
]

# Filter to the columns that exist, then rename
existing = [c for c in keep if c in pov_df.columns]

# Dictionary for renaming columns
renamed = {
    "S1701_C01_001E": "Total Population",
    "S1701_C02_001E": "Population Below Poverty",     
    "S1701_C03_001E": "Population Below Poverty Percent",      
    "S1701_C01_023E": "Adults 25 and Older Less Than High School Graduate",       
    "S1701_C01_026E": "Adults 25 and Older with Bachelor degree or higher",              
}
# Create the output DataFrame with existing columns and renamed
pov_df = pov_df[existing].rename(columns=renamed)
```

```python
# Calculating percent columns (value / Population Below Poverty) * 100
percent_cols = ["Adults 25 and Older Less Than High School Graduate Percent", "Adults 25 and Older with Bachelor's degree or higher Percent"]
existing_percent_cols = [col for col in percent_cols if col in pov_df.columns]
other_cols = [col for col in pov_df.columns if col not in existing_percent_cols]
pov_df = pov_df[existing_percent_cols + other_cols]


def insert_after(df, after_col, new_col, values):
    cols = list(df.columns)
    if after_col in cols:
        idx = cols.index(after_col) + 1
 
        if new_col in cols:
            cols.remove(new_col)
        cols = cols[:idx] + [new_col] + cols[idx:]
        df[new_col] = values
        df = df[cols]
    else:
        df[new_col] = values
    return df

# Incomplete High School Graduates (Percent)
if "Adults 25 and Older Less Than High School Graduate" in pov_df.columns and "Total Population" in pov_df.columns:
    percent = (
        pd.to_numeric(pov_df["Adults 25 and Older Less Than High School Graduate"], errors="coerce") /
        pd.to_numeric(pov_df["Total Population"], errors="coerce")
    ) * 100
    percent = percent.round(1)
    pov_df = insert_after(
        pov_df,
        "Adults 25 and Older Less Than High School Graduate",
        "Adults 25 and Older Less Than High School Graduate Percent",
        percent
    )

# Bachelor degree or higher (Percent)
if "Adults 25 and Older with Bachelor degree or higher" in pov_df.columns and "Total Population" in pov_df.columns:
    percent = (
        pd.to_numeric(pov_df["Adults 25 and Older with Bachelor degree or higher"], errors="coerce") /
        pd.to_numeric(pov_df["Total Population"], errors="coerce")
    ) * 100
    percent = percent.round(1)
    pov_df = insert_after(
        pov_df,
        "Adults 25 and Older with Bachelor degree or higher",
        "Adults 25 and Older with Bachelor degree or higher Percent",
        percent
    )
```

```python
# Handle missing values
pov_df = pov_df.replace('(X)', None)

pov_df.columns = [col.lower().replace(' ', '_') for col in pov_df.columns]

# Extract only digits from year column, drop bad text
pov_df["year"] = pov_df["year"].astype(str).str.extract(r"(\d+)")

# Convert to integer
pov_df["year"] = pov_df["year"].astype(int)

# Build mapping dynamically for 2-digit years
year_map = {y: 2000 + y for y in pov_df["year"].unique() if y < 100}

# Apply replacement
pov_df["year"] = pov_df["year"].replace(year_map)

print(year_map)                # e.g. {15: 2015, 16: 2016}
print(pov_df["year"].unique()) # should show full 4-digit years
```

# Combining Education, Income, Poverty and Investment

```python
print("df_poverty columns:", df_poverty.columns.tolist())
print("df_investment columns:", df_investment.columns.tolist())
print("df_income columns:", df_income.columns.tolist())
print("df_education columns:", df_education.columns.tolist())
```

```python
from functools import reduce
```

```python
def normalize(df, fips_col="county_fips"):
    df = df.copy()
    if fips_col in df.columns:
        df[fips_col] = (
            df[fips_col].astype(str).str.extract(r"(\d{5})", expand=False).str.zfill(5)
        )
    if "year" in df.columns:
        df["year"] = pd.to_numeric(df["year"], errors="coerce").astype("Int64")
    return df

def strip_meta(df):
    """Drop duplicate meta columns so we keep state/county only once (from poverty)."""
    return df.drop(columns=[c for c in ["state", "county", "geo_id"] if c in df.columns])

for k in df:
    df[k] = df[k].drop_duplicates(ignore_index=True) 
```

```python
# Normalize
df_poverty    = normalize(df_poverty)
df_investment = normalize(df_investment)
df_income     = normalize(df_income)
df_education  = normalize(df_education)
```

```python
# Build the (county_fips, year) intersection
dfs_with_year = [df_poverty, df_investment, df_income, df_education]

key_df = reduce(
    lambda a, b: a.merge(b, on=["county_fips", "year"], how="inner"),
    [d[["county_fips", "year"]].dropna().drop_duplicates() for d in dfs_with_year]
)
```

```python
# Filter each DF to the key intersection
dfs_filtered = [key_df.merge(d, on=["county_fips", "year"], how="inner") for d in dfs_with_year]
```

```python
# Now they all have identical (county_fips, year) rows
f_poverty, f_investment, f_income, f_education = dfs_filtered
```

```python
# Drop duplicate meta cols before final merge
to_merge = [strip_meta(d) for d in [f_poverty, f_investment, f_income, f_education]]
```

```python
# Inner-merge all time-series on (county_fips, year)
df_merged = reduce(
    lambda left, right: left.merge(right, on=["county_fips", "year"], how="inner"),
    to_merge
)
```

```python
df_merged = df_merged.merge(
    f_poverty[["county_fips", "year", "state", "county"]].drop_duplicates(),
    on=["county_fips", "year"],
    how="left"
)
```

```python
id_cols = ["county_fips", "state", "county", "year"]
df_final_502 = df_merged[id_cols + [c for c in df_merged.columns if c not in id_cols]].drop_duplicates()
```

```python
df_final_502 = df_final_502.drop_duplicates(ignore_index=True)
df_final_502 = df_final_502.drop_duplicates(subset=["county_fips", "year"], keep="first").reset_index(drop=True)
```

```python
# Add Investment_dollars_per_capita column
df_final_502["investment_dollars_per_capita"] = (
    (df_final_502["total_investment_dollars"].astype(float) / df_final_502["total_population"].astype(float)).round(2)
)
```
